```python
# Calculate the slope, x-intercept and y-intercept of y = 2x -2
slope = 2
y_intercept = -2
x_intercept = -y_intercept / slope
print(f"The slope of the line is: {slope}")
print(f"The y-intercept of the line is: {y_intercept}")
print(f"The x-intercept of the line is: {x_intercept}")
```

```python
#Slope is (m = y2-y1/x2-x1). Find the slope and Euclidean distance between point (2, 2) and point (6,10)
x1, y1 = 2, 2
x2, y2 = 6, 10
slope = (y2 - y1) / (x2 - x1)
euclidean_distance = ((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5
print(f"The slope of the line is: {slope}")
print(f"The Euclidean distance between the points is: {euclidean_distance}")
```

```python
#Compare the slopes in the two questions above.
if slope == 2:
    print("The slopes are equal.")
else:
    print("The slopes are not equal.")
```

```python
# Check if the floor division of 7 by 3 is equal to the int converted value of 2.7.
floor_division = 7 // 3
int_converted = int(2.7)
print(f"Floor division of 7 by 3: {floor_division}")
print(f"Int converted value of 2.7: {int_converted}")
print(f"Are they equal? {floor_division == int_converted}")
```
