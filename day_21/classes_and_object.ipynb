```python
import statistics
from collections import Counter


class Statistics:
    def __init__(self, data):
        if not data:
            raise ValueError("Data cannot be empty")
        if not all(isinstance(value, (int, float)) for value in data):
            raise TypeError("Data must contain only numbers")
        self.data = list(data)

    def count(self):
        return len(self.data)

    def sum(self):
        return sum(self.data)

    def min(self):
        return min(self.data)

    def max(self):
        return max(self.data)

    def range(self):
        return self.max() - self.min()

    def mean(self):
        value = statistics.mean(self.data)
        return round(value)

    def median(self):
        value = statistics.median(self.data)
        return int(value) if value.is_integer() else round(value, 2)

    def mode(self):
        frequency = Counter(self.data)
        highest_count = max(frequency.values())
        mode_value = min(value for value, count in frequency.items() if count == highest_count)
        return mode_value, highest_count

    def var(self):
        # Population variance matches the exercise's 17.5 result.
        return round(statistics.pvariance(self.data), 1)

    def std(self):
        return round(statistics.pstdev(self.data), 1)

    def percentile(self, percentile):
        if not 0 <= percentile <= 100:
            raise ValueError("Percentile must be between 0 and 100")
        ordered = sorted(self.data)
        position = (len(ordered) - 1) * percentile / 100
        lower = int(position)
        upper = min(lower + 1, len(ordered) - 1)
        fraction = position - lower
        return ordered[lower] * (1 - fraction) + ordered[upper] * fraction

    def freq_dist(self):
        frequency = Counter(self.data)
        total = len(self.data)
        distribution = [
            (round(count / total * 100, 1), value)
            for value, count in sorted(frequency.items(), key=lambda item: (-item[1], -item[0]))
        ]
        return distribution

    def describe(self):
        mode_value, mode_count = self.mode()
        print("Count:", self.count())
        print("Sum:", self.sum())
        print("Min:", self.min())
        print("Max:", self.max())
        print("Range:", self.range())
        print("Mean:", self.mean())
        print("Median:", self.median())
        print("Mode:", (mode_value, mode_count))
        print("Variance:", self.var())
        print("Standard Deviation:", self.std())
        print("Frequency Distribution:", self.freq_dist())


ages = [31, 26, 34, 37, 27, 26, 32, 32, 26, 27, 27, 24, 32, 33, 27, 25, 26, 38, 37, 31, 34, 24, 33, 29, 26]
data = Statistics(ages)

print("Count:", data.count())
print("Sum:", data.sum())
print("Min:", data.min())
print("Max:", data.max())
print("Range:", data.range())
print("Mean:", data.mean())
print("Median:", data.median())
print("Mode:", {"mode": data.mode()[0], "count": data.mode()[1]})
print("Standard Deviation:", data.std())
print("Variance:", data.var())
print("Frequency Distribution:", data.freq_dist())

print("\nDescribe:")
data.describe()
print("25th percentile:", data.percentile(25))
```

```python
class PersonAccount:
    def __init__(self, firstname, lastname):
        self.firstname = firstname
        self.lastname = lastname
        self.incomes = set()
        self.expenses = set()

    def add_income(self, description, amount):
        if amount < 0:
            raise ValueError("Income amount cannot be negative")
        self.incomes.add((description, amount))

    def add_expense(self, description, amount):
        if amount < 0:
            raise ValueError("Expense amount cannot be negative")
        self.expenses.add((description, amount))

    def total_income(self):
        return sum(amount for _, amount in self.incomes)

    def total_expense(self):
        return sum(amount for _, amount in self.expenses)

    def account_balance(self):
        return self.total_income() - self.total_expense()

    def account_info(self):
        print(f"Account holder: {self.firstname} {self.lastname}")
        print("Incomes:", sorted(self.incomes))
        print("Expenses:", sorted(self.expenses))
        print("Total income:", self.total_income())
        print("Total expense:", self.total_expense())
        print("Account balance:", self.account_balance())


account = PersonAccount("Ahmad", "Khamis")
account.add_income("Salary", 3000)
account.add_income("Freelance", 500)
account.add_expense("Rent", 900)
account.add_expense("Food", 350)

account.account_info()
```
