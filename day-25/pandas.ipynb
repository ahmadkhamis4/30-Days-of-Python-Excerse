```python
from pathlib import Path
import pandas as pd

csv_path = Path("data") / "hacker_news.csv"
if not csv_path.exists():
    csv_path = Path("..") / "data" / "hacker_news.csv"
if not csv_path.exists():
    raise FileNotFoundError(f"Could not find hacker_news.csv from {Path.cwd()}")

hacker_news_df = pd.read_csv(csv_path)
print(f"Loaded {len(hacker_news_df):,} rows and {len(hacker_news_df.columns)} columns from {csv_path.resolve()}")
```

```python
# Show the first five rows.
hacker_news_df.head()
```

```python
# Show the last five rows.
hacker_news_df.tail()
```

```python
# Select the title column as a pandas Series.
titles = hacker_news_df["title"]
print("Type:", type(titles))
titles
```

```python
# Report the number of rows and columns.
rows, columns = hacker_news_df.shape
print("Rows:", rows)
print("Columns:", columns)
```

```python
# Filter titles containing Python, ignoring case and missing titles.
python_titles = hacker_news_df.loc[
    hacker_news_df["title"].str.contains("python", case=False, na=False),
    "title",
]
print(f"Found {len(python_titles)} titles containing Python.")
python_titles
```

```python
# Filter titles containing JavaScript, ignoring case and missing titles.
javascript_titles = hacker_news_df.loc[
    hacker_news_df["title"].str.contains("JavaScript", case=False, na=False),
    "title",
]
print(f"Found {len(javascript_titles)} titles containing JavaScript.")
javascript_titles
```

```python
# Explore column types, missing values, numeric distributions, and active authors.
print("Dataset shape:", hacker_news_df.shape)

print("\nColumn data types:")
print(hacker_news_df.dtypes)

print("\nMissing values per column:")
print(hacker_news_df.isna().sum())

print("\nNumeric column summary:")
print(hacker_news_df.describe())

print("\nTop 10 authors by number of posts:")
hacker_news_df["author"].value_counts().head(10)
```
