```python
#Exercise: Level 2
#The following is a list of 10 students ages:
ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24]
#Sort the list
ages.sort()
print(ages)
#Find the min and max age
min_age = min(ages)
max_age = max(ages)
print(f"Minimum age: {min_age}")
print(f"Maximum age: {max_age}")
#Find the median age (one middle item or two middle items divided by two)
n = len(ages)
if n % 2 == 0:
    median_age = (ages[n//2 - 1] + ages[n//2]) / 2
else:
    median_age = ages[n//2]
print(f"Median age: {median_age}")
#Find the average age (sum of all items divided by their number)
average_age = sum(ages) / n
print(f"Average age: {average_age}")
#Find the range of the ages (max minus min)
age_range = max_age - min_age
print(f"Range of ages: {age_range}")
#Compare the value of (min - average) and (max - average), use abs() method
min_diff = abs(min_age - average_age)
max_diff = abs(max_age - average_age)
print(f"Absolute difference between min and average: {min_diff}")
print(f"Absolute difference between max and average: {max_diff}")

#Find the middle country(ies) in the countries list
#Divide the countries list into two equal lists if it is even if not one more country for the first half.
#['China', 'Russia', 'USA', 'Finland', 'Sweden', 'Norway', 'Denmark']. Unpack the first three countries and the rest as scandic countries.
countries_list = ['China', 'Russia', 'USA', 'Finland', 'Sweden', 'Norway', 'Denmark']
middle_index = len(countries_list) // 2
first_half = countries_list[:middle_index]
second_half = countries_list[middle_index:]
print(f"First half: {first_half}")
print(f"Second half: {second_half}")
```
